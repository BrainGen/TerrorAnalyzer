```python
import pandas as pd
import numpy as np


terrorism = pd.read_csv(r"Global Terrorism Dataset\globalterrorismdb_0718dist.csv", encoding="Windows-1252")
```

```python
replace_dict = {
    "unknown":pd.NA,
    "maoists":"communist party of india - maoist (cpi-maoist)",
    "palestinian extremists":"palestine liberation organization (plo)",
    "palestinians": "palestine liberation organization (plo)", 
    "muslim militants":"muslim extremists",
    "muslim separatists":"muslim extremists",
    "muslim fundamentalists":"muslim extremists",
    "muslim rebels":"muslim extremists",
    "muslims":"muslim extremists",
    "islamic state of iraq (isi)":"islamic state of iraq and the levant (isil)" 
}
```

```python
#Run this cell only once. The conversion of unknown values to NaN leads to them being dropped on a second run.

terrorism["gname"] =(
    terrorism["gname"]
    .str.strip()
    .str.lower()
    .replace(to_replace=replace_dict)
) 


ranks = (
    terrorism
    .groupby("gname")["eventid"]
    .count()
    .sort_values(ascending=False)
)

ranks = ranks[ranks>30]
terrorism = terrorism[terrorism["gname"].isin(ranks.index)]

with open(r"ranks.txt", "w") as file_rank:
    file_rank.write(ranks.to_string())

```

```python
default_col = [
    "target",
    "ransom",
    "targ",
    "corp",
    "natlty",
    "guncertain",
    "claimmode",
    "weap",
    "ishostkid",
    "scite",
    "hostkidoutcome",
    "prop",
    "nperp"
]
additional_columns = [
    "attacktype2_txt",
    "attacktype2",
    "attacktype3",
    "attacktype3_txt",
    "gsubname2",
    "gsubname3",
    "gname2",
    "gname3",
    "dbsource",
    "addnotes",
    "nkillus",
    "nwoundus",
    "nkillter",
    "nwoundte",
    "location" 
]
```

```python

collist = [col for col in terrorism.columns if any(w in col.lower() for w in default_col)] + additional_columns
    

terrorism =terrorism.drop(columns=collist) 

terrorism = terrorism.drop(columns= [col for col in terrorism.columns if (terrorism[col].isna().mean())>0.85])
terrorism.columns
```

```python
terrorism.isna().sum()
```

```python
#Date Parsing code
def date_str(df):
    return (df.iyear.astype(str)+ "/" + df.imonth.astype(str) + "/" + df.iday.astype(str))

terrorism["date"] = (
    terrorism
    .pipe(date_str)
    .pipe(pd.to_datetime, format=f"%Y/%#m/%#d", errors="coerce")
)

terrorism =(
    terrorism
    .dropna(subset="date")
    .drop(columns=["imonth", "iday"])
) 

```

```python
cities = pd.read_csv(f"Global Terrorism Dataset/worldcitiespop.csv")

cities["City"] =(
    cities["City"]
    .str.lower()
    .str.strip()
) 
cities = cities.drop_duplicates(subset=["City"])

terrorism["city"] =(
    terrorism["city"]
    .str.lower()
    .str.strip()
) 

terrorism = pd.merge(terrorism, cities, how="left", left_on="city", right_on="City")

terrorism["latitude"] = terrorism.latitude.fillna(terrorism.Latitude)
terrorism["longitude"] = terrorism.longitude.fillna(terrorism.Longitude)
```

```python
drops = ["Country","City","AccentCity","Region","Population","Latitude","Longitude"]
terrorism = terrorism.drop(columns= drops)
```

```python
analytics =(
    terrorism
    .groupby(["gname", "country_txt"], as_index=False)
    .filter(lambda x:x["nkill"].sum() > 20 )
    .groupby(["gname", "country_txt"], as_index=False)
    .count()
    .sort_values("nkill", ascending=False)
    .sort_values(by="country_txt", kind="stable")
    .set_index(["country_txt", "gname"])
    
)
analytics_txt = analytics[["nkill","nwound"]].to_string()
with open(r"analytics.txt", "w") as f:
    f.write(analytics_txt)
    f.close()
```

```python
#Cleaning of Uppsala Uni Dataset


georef = pd.read_csv("Global Terrorism Dataset\Georef.csv\GEDEvent_v26_1.csv")

filtered = georef.loc[:,["year", "conflict_name", "country", "country_id", "region", "deaths_civilians", "best"]]

filtered.country = (
    filtered.country
    .str.lower()
    .str.split("(")
    .str[0]
    .str.strip()
    
)

civilians_deaths = (
    filtered
    .groupby("country")
    .sum()
)
civilians_deaths.sort_values("deaths_civilians",ascending=False)
```

```python
civilians_deaths = (
    filtered
    .groupby(["country", "year"], as_index=False)
    .sum()
)
def reindex_list(df):
    reindexed = (
        df
        .groupby("country", as_index=False)
        .sum()
        .sort_values("deaths_civilians", ascending = False)
        .country
    )
    return reindexed

civilians_deaths = (
    civilians_deaths
    .sort_values("deaths_civilians", ascending=False)
    .sort_values("country", kind="stable")
    .set_index(["country", "year"])
    .pipe(lambda d:d.reindex(reindex_list(civilians_deaths), level="country"))
    .reset_index()
    .loc[:,["country", "year", "deaths_civilians"]]
)
terrorism

terrorism = (
    terrorism
    .drop(columns = "country")
    .rename(columns={"iyear":"year", "country_txt":"country"})
)

terrorism["country"] = (
    terrorism["country"]
    .str.lower()
    .str.strip()
)




terrorism_and_atrocity = pd.merge(left=terrorism, right=civilians_deaths, on=["country", "year"], how="left")
```

```python
terrorism_and_atrocity = terrorism_and_atrocity.loc[:, ["year", "country", "date", "region_txt", "provstate", "city", "nkill", "nwound", "deaths_civilians",]]
```

```python
econ = pd.read_excel(r"Global Terrorism Dataset/hdr-data.xlsx").loc[:, ["country", "year", "value"]]
non_countries = [
    'very high human development',
    'high human development',
    'medium human development',
    'low human development',
    'arab states',
    'east asia and the pacific',
    'europe and central asia',
    'latin america and the caribbean',
    'south asia',
    'sub-saharan africa',
    'world'
]
econ["country"] = (
    econ["country"]
    .str.strip()
    .str.lower()
    

)
econ["country"] = econ["country"][~econ["country"].isin(non_countries)]

econ = econ.dropna()

without_inequality = pd.merge(left=terrorism_and_atrocity, right=econ, on=["country", "year"])
without_inequality.isna().sum()
```

```python
import itertools
inequality_dataset = pd.read_csv("Global Terrorism Dataset/EPR-2018.1.1.csv")
inequality_dataset
def expand_years(df):
    n = df["to"] - df["from"] +1
    idx = np.repeat(df.index.values, n)
    expanded_df = df.loc[idx].copy()
    expanded_df["year"] = np.fromiter(
        itertools.chain.from_iterable(
            np.arange(row["from"], row["to"]+1)
            for _, row in df.iterrows()
        ),
        dtype=int
    )
    return expanded_df

inequality_dataset = expand_years(inequality_dataset)
inequality_dataset["status"] = (
    inequality_dataset["status"]
    .str.lower()
    .str.strip()
    .replace(to_replace={"monopoly": "8", "dominant":"7", "senior partner":"6", "junior partner":"5", "self-exclusion":"4", "powerless":"3", "irrelevant":"3", "discriminated":"2", "state collapse":"1"})
    .astype(int)
)
inequality_dataset["e2c"] = inequality_dataset["status"]/inequality_dataset["size"]
```

```python
e2c_grouped = inequality_dataset.groupby(["statename", "year"])["e2c"]
inequality_dataset["weighted_squared_power"] = (inequality_dataset["e2c"]/e2c_grouped.transform("sum"))**2
metric = inequality_dataset.groupby(["statename", "year"], as_index=False)["weighted_squared_power"].sum()
no_of_ethnic_groups = inequality_dataset.groupby(["statename", "year"], as_index=False).count()
countries = no_of_ethnic_groups[no_of_ethnic_groups["e2c"]==1]["statename"]
metric[metric["statename"].isin(countries)] = 0
metric = metric[["statename", "year", "weighted_squared_power"]]
metric.isna().sum()
metric = metric.rename(columns={"statename":"country", "weighted_squared_power":"inequality"})
metric["country"] = (
    metric["country"]
    .str.strip()
    .str.lower()
)
metric.country.unique()
with_inequality = pd.merge(left=without_inequality, right=metric, how='left')
with_inequality.isna().sum()
```

```python
#Chronological Lag Features
aggregate_data = with_inequality.groupby(["country", "year"], as_index=False)[["nkill", "nwound", "deaths_civilians", "value", "inequality"]].agg({"nkill":"sum", "nwound":"sum", "value":"mean", "inequality":"mean", "deaths_civilians":"mean"})

aggregate_data = aggregate_data.ffill(limit=10)
aggregate_data = aggregate_data.bfill(limit=10)
aggregate_data = aggregate_data.fillna(aggregate_data.mean(numeric_only=True))
aggregate_data.isna().sum()

aggregate_data["nkilldelay1"] = aggregate_data.groupby("country")["nkill"].shift(1)
aggregate_data["nkilldelay2"] = aggregate_data.groupby("country")["nkill"].shift(2)
aggregate_data["nwounddelay1"] = aggregate_data.groupby("country")["nwound"].shift(1)
aggregate_data["nwounddelay2"] = aggregate_data.groupby("country")["nwound"].shift(2)

aggregate_data = aggregate_data.dropna()
```

```python
aggregate_data.to_csv("final.csv",index=False )
```
